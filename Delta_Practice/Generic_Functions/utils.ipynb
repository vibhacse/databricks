### Generic Read/Write File Functions

```python
def read_csv(session,path,header=True,inferSchema=True):
  return session.read.csv(path,header=header,inferSchema=inferSchema)

def write_delta_file(sparkdf,format,path,mode='overwrite'):
  return sparkdf.write.format(format).mode(mode).save(path) 
```

```python
#To display the description of the delta table
def describe_delta_table(delta_table_name):
  return spark.sql(f'desc history {delta_table_name};')
```
