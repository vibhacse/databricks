```python
df1=spark.readStream.format("cloudFiles")\
    .option("cloudFiles.Format","csv")\
    .option("cloudFiles.schemaEvolutionMode","addNewColumns")\
    .option("cloudFiles.maxFilesPerTrigger",1)\
    .option("cloudFiles.inferColumnTypes",True)\
    .option("checkpointLocation", "/Volumes/prodcatalog/logistics/bronze/streamwriter_delta/_checkpoint")\
    .option("cloudFiles.schemaLocation", "/Volumes/prodcatalog/logistics/bronze/streamwriter/_schema")\
    .option("header",True)\
.load(f"/Volumes/prodcatalog/logistics/source/")
```

### To perform schema evolution, we have to use the below properties:<br>
**Read side:**<br>
.option("cloudFiles.schemaEvolutionMode","addNewColumns")<br>
**Write side:**<br>
.option("mergeSchema", "true")<br>

```python
df1.writeStream.trigger(availableNow=True)\
    .option("checkpointLocation", "/Volumes/prodcatalog/logistics/bronze/streamwriter/_checkpoint")\
    .option("cloudFiles.schemaLocation", "/Volumes/prodcatalog/logistics/bronze/streamwriter/_schema")\
    .option("mergeSchema",True)\
.start(f"/Volumes/prodcatalog/logistics/bronze/streamwriter/")
```

```python
spark.read.format("delta").load("/Volumes/prodcatalog/logistics/bronze/streamwriter").show()
```

```python
#to store the raw data into delta table
df1.writeStream.trigger(availableNow=True)\
    .option("checkpointLocation", "/Volumes/prodcatalog/logistics/bronze/streamwriter_delta/_checkpoint")\
    .option("cloudFiles.schemaLocation", "/Volumes/prodcatalog/logistics/bronze/streamwriter/_schema")\
    .option("mergeSchema",True)\
.toTable("prodcatalog.logistics.bronze")
```
