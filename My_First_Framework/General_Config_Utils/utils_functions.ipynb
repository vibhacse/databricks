# **Generic Functions**

```python
from pyspark.sql.session import SparkSession

def get_spark_session(appname="some name for the project"):
  spark = SparkSession.builder.appName(appname).getOrCreate()
  return spark
```

```python
#function to read different type of files

def read_files(spark,filetype,file_path,seperator=',',infer_schema=True,Header=True,mline=True):
    if filetype == "text":
        return spark.read.format(filetype).options(path=file_path,sep=seperator,inferSchema=infer_schema,header=Header)
    if filetype == "csv":
        return spark.read.format(filetype).options(path=file_path,inferSchema=infer_schema,header=Header)
    elif filetype == "json":
        return spark.read.format(filetype).options(path=file_path,multiline=mline)
    else:
        return None
        
```
