```python
dbutils.widgets.text("catalog","defaultcatalog")
dbutils.widgets.text("schema","defaultschema")
dbutils.widgets.text("table","defaulttable")
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
table = dbutils.widgets.get("table")
print(catalog)
print(schema)
print(table)
```

```python
rawdf=spark.read.csv(path="/Volumes/prodcatalog/logistics/source/logistics_source2.csv", header=True, inferSchema= True)
rawdf.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.{table}")
```
